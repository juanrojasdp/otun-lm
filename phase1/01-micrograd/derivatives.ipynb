```python
def df(x):
    return 6*x - 4

def new_x(x,lr):
    return x - df(x)*lr

x=0
lr = 0.1
for i in range(20):
    x_new = new_x(x,lr)
    print(x_new)
    x = x_new

```

```python
def gradient_descent(lr, steps=20, x0=0.0):
    x = x0
    for i in range(steps):
        x = x - df(x)*lr
        print(x)
    return x

gradient_descent(0.1)
```

```python
def numerical_derivative(fn, x, h):
    return (fn(x + h) - fn(x))/h

numerical_derivative(f,3,1e-1)
```

```python
for h in [1e-1, 1e-3, 1e-5, 1e-8, 1e-12, 1e-15, 1e-17]:
    print(numerical_derivative(f,2,h))
```
