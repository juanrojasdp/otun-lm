```python
def df(x):
    return 6*x - 4

def new_x(x,lr):
    return x - df(x)*lr

x=0
lr = 0.1
for i in range(20):
    x_new = new_x(x,lr)
    print(x_new)
    x = x_new

```

```python
def gradient_descent(lr, steps=20, x0=0.0):
    x = x0
    for i in range(steps):
        x = x - df(x)*lr
        print(x)
    return x

gradient_descent(0.1)
```
